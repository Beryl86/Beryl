## Tuple

Tuples are essentially lists. Tuples are used to group objects of different types. In contrast to lists, tuples cannot be modified after creating them (i.e. they are *immutable*). 

We can rewrite our first list as a tuple by using round brackets or by invoking *tuple*.

```python
first_tuple = (2, 4, 6, 8, 10)
first_tuple
```

```python
type(first_tuple)
```

```python
tup = tuple(['hello', 2, 4, True])
tup
```

If we try to change the fourth element of tuple, we get an error message:

```python
tup[3] = "False"
```

Indeed, once the tuple is created, it is not possible to modify the object is stored in each slot. 

There are only two methods for tuples of numbers, since the size and content of a tuple cannot be modified. 
The first method is *count*, which counts how many times a specific value is present in the tuple.

```python
num_tup = (2, 4, 6, 8, 8, 8, 10)
num_tup.count(8)
```

The second method is *index*, which tells us at what index the given number first appears in the tuple.

```python
num_tup.index(8)
```

Why/when would you use a tuple instead of a list? 
* Tuples are **faster**. This can become relevant if you need to iterate over a large number of items.
* Tuples are **safer**. As they are immutable, they are basically write-protected, which can be important for a complicated project.
* Tuples can be used as **keys** in dictionary objects (see later).

#  Conditional and Iterative Processing

Python has several built-in keywords for conditional logic and loops. 

## Conditional Processing: if, elif, and else 

**The *if* Statement**

The *if* statement is one of the most important control flow statement types. We use the *if* statement combined with a condition/statement that returns True of False. If that condition is True, then the code in the block that follows is evaluated. If the condition is False, then the code is not evaluated.

```python
age = 20

if age >= 18:
    print("You are an adult.")
```

```python
x = -5
if x < 0:
    print("This is a negative number")
```

**The *if-else* Statement**

Optionally, you can provide code that has to be evaluated when the condition is false using *else*.

```python
age = 45

if age >= 18:
    print("You are an adult.")
else:
    print("You are a minor.")
```

```python
x = -100
if x < 0:
    print("This is a negative number")
else:
    print("This is a nonnegative number")
```

**The *elif* Statement**

Optionally, you can use one or more *elif* blocks, but the final statement has to be *else*. The conditions are evaluated in order. If one of them is True, the subsequent code is evaluated and no further *elif* and *else* blocks will be evaluated.  

```python
x = 50
if x > 100:
    print("this number is big")
elif x > 10:
    print("this number is not big and not small")
else:
    print("this number is small")
```

```python
score = 60

if score >= 80:
    print("Grade: A")
elif score >= 70:
    print("Grade: B")
elif score >= 60:
    print("Grade: C")
elif score >= 50:
    print("Grade: D")
else:
    print("Grade: F")
```

###  *for* loops

Why use loops? If you want to avoid to write the same code over and over again, you might want to use a *for* loop. Below you can find one of the simplest examples of using a *for* loop.

```python
for i in range(5):
    print(i)
```

For example, if we want to access the average cholesterol values of all patients in the list that we've just created, we can use the following code: 

```python
tot_chol = 0.0
for p in patient:
    chol = p[2]
    tot_chol += chol
tot_chol
```

```python
avg_chol = tot_chol / len(patient)
avg_chol
```

The average cholosterol level in our patient data is 205.2. Thanks to using a *for* loop, we saved time in computing the mean. However, what we really have done here? 

First we create the variable that will contain the sum of all cholesterol levels and we initialize it to zero. Then we use a *for loop*. The syntax says that for every *p* in patient, the cholesterol values (in position 2) are accessed and then the total chol level will be incremented by the individual cholesterol levels within the loop.  

Two remarks about the code:
* Using the letter *p* is arbitrary. You could use any other letter or word. However, using the first letter of the list you are looping is the convention. 

* The operator += is the addition assignment operator in Python. It basically adds the value of the second variable to the first one, and assigns the result to the first variable: A += B is the same as to A = A + B.

###  *while* loops

A *while* loop repeats a block of code **as long as** a condition is true.

It is like a repeated *if* statement. It specifies a condition and a block of code that is executed over and over again as long as the condition is *True*.

```python
count = 5
while count < 9:
    print ('The count is:', count)
    count = count + 1
    
print ("Good bye!")
```

```python
count = 1
while count <= 5:
       print (f" Count is { count }")
       count = count + 1
print (" Done !")
```

Remember that *while* loops should always end at some point! Pay attention not to create infinite loops!

A function is defined by the keyword *def* and it returns the object in the *return* statement. Multiple *return* statements are allowed. Here is an example:

### Functions that Return a Value

A function can also **return** a result using the *return* keyword:

```python
def add(a, b):
    result = a + b
    return result

sum1 = add(3, 5)
sum2 = add(10, 20)

print(sum1)   # 8
print(sum2)   # 30
```

```python
def my_funct(x, y, z = 2):
    if z > 1:
        return z * (x + y)
    else:
        return z / (x + y)     
```

The third argument has been given a default value of 2. If we do not specify a third argument, the default is used.

```python
output = my_funct(2, 3, 0.5)
output
```

```python
my_funct(2, 3)
```

When a function is called, arguments can be provided as **positional arguments** and **keyword arguments**. 

In the following function call, all arguments were provided as positional arguments. The right value is attached to the right variable name based on position: the first one is x, the second one is y, and the third one is z.



```python
my_funct(2, 3, 0.5)
```

In the following function call, all arguments were provided as keyword arguments. Note that the order does not matter. 

```python
my_funct(y = 2, x = 3)
```

Even though I do not recommend it, you can use a combination of positional and keyword arguments. There are some restrictions on what you can do, however, as illustrated below. 

* All positional arguments need to preceed all keyword arguments.
* The positional arguments need to be in the right position.

```python
my_funct(x = 3, y = 2)
```

```python
my_funct(3, y = 2)
```

## Two-sample t-test

One of the most widely known basic bivariate statistical tests is the Student's t-test, which tests whether or not there is a statistically significant difference between the means in two groups. 

We will use the iris dataset here, which includes information on the sepal and petal length and width of different flower species. 

```python
df = pd.read_csv("Iris_Data.csv")
df.head() #check the first rows of your data
```

It is always a good idea to first have a look at your data using some descriptives. The *.describe()* function is one way of doing this.

```python
# Descriptive statistics
df.groupby("species")["sepal_width"].describe()
```

Let us focus on Iris-setosa and Iris-versicolor. There seems to be a difference between the average sepal width and the standard deviations are small. But is this difference significant?

We create two *DataFrames*, one with the setosa observations, the other with the versicolor observations.

```python
setosa = df[(df['species'] == 'Iris-setosa')] # dataframe only containing setosa flowers
setosa.reset_index(inplace = True) # create an index variable
setosa.head()
```

```python
versicolor = df[(df['species'] == 'Iris-versicolor')] # dataframe only containing versicolor flowers
versicolor.reset_index(inplace= True) # create an index variable
versicolor.head()
```

### Assumption checks
#### Homogeneity of variances

Levene's test for homogeneity of variance is widely used. 

```python
stats.levene(setosa['sepal_width'], versicolor['sepal_width'])
```

The p-value is 0.42, which means that according to Levene's test there is no reason to suspect a difference in variances between the two groups. We can proceed.

#### Normally distributed

Checking for normality can be done visually (e.g. q-q plot or histogram) or using a statistical test (e.g. Shapiro-Wilk test to formally test for normality).

```python
stats.probplot(setosa['sepal_width'], plot = plt)
plt.title('Sepal Width qq-Plot for setose')
```

```python
stats.probplot(versicolor['sepal_width'], plot = plt)
plt.title('Sepal Width qq-Plot for versicolor')
```

```python
plt.hist(setosa['sepal_width'], bins = 10)
plt.title("Sepal Width Histogram for setosa")
```

```python
plt.hist(versicolor['sepal_width'], bins = 10)
plt.title("Sepal Width Histogram for versicolor")
```

Alternatively, we can use a Shapiro-Wilk test.

```python
stats.shapiro(setosa['sepal_width'])
```

```python
stats.shapiro(versicolor['sepal_width'])
```

The first value is the W test statistic and the second one the p-value. Given the data, we cannot conclude that there is a significant deviation from normality in either species.

###  Two sample t-test

```python
stats.ttest_ind(setosa['sepal_width'], versicolor['sepal_width'])
```

```python
out = stats.ttest_ind(setosa['sepal_width'], versicolor['sepal_width'])
print('The p-value is %.20f.' %out.pvalue) # to control the number of decimals
```

We conclude that Iris-setosa’s average sepal width (M = 3.418, SD = 0.381) is significantly larger than Iris-versicolor (M = 2.770, SD = 0.314). A two-sample t-test was used to test for a difference.

## Chi square test

Another well known test is the chi-square test, which is used to test for independence of two categorical variables. 

Consider the *workinghours* dataset, which contains, for a random sample of a population, the sex of the person and the hours they work per week (categorical). We will be looking into the relationship between these two variables. 

```python
import pandas as pd
workhour_by_sex = pd.read_csv('workinghours.csv')
workhour_by_sex = workhour_by_sex[['sex', 'hours_per_week_categories']]
workhour_by_sex.head()
```

Let us check the division of workhours and of  sex.

```python
workhour_by_sex['hours_per_week_categories'].value_counts()
```

```python
workhour_by_sex['sex'].value_counts()
```

```python
sns.catplot(data = workhour_by_sex,
            x = "sex",
            kind = "count")
```

```python
sns.catplot(data = workhour_by_sex,
            x = "hours_per_week_categories",
            kind = "count",
            col = "sex",
            order = sorted(workhour_by_sex['hours_per_week_categories'].unique()))
```

### Constructing the contingency table

A first step is to construct the frequency count table or the contingency table. This is the crosstable with the observed frequencies, used to calculate the value of the test statistic.

```python
contingency_table = pd.crosstab(
    workhour_by_sex['sex'],
    workhour_by_sex['hours_per_week_categories'],
    margins = True)
contingency_table
```

We can visualize it with a stacked barplot.

```python
# Assigns the frequency values
femalecount = contingency_table.iloc[0][0:6].values
malecount = contingency_table.iloc[1][0:6].values
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plots the bar chart
fig = plt.figure(figsize = (10, 5))
sns.set(font_scale = 1.8)
categories = ["0-9","10-19", "20-29", "30-39", "40-49", "50+"]
p1 = plt.bar(categories, malecount, .55, color = '#d62728')
p2 = plt.bar(categories, femalecount, 0.55, bottom = malecount)
plt.legend((p2[0], p1[0]), ('Male', 'Female'))
plt.xlabel('Hours per Week Worked')
plt.ylabel('Count')
plt.show()
```

### Calculating the Chi square test statistic


We can calculate the chi-squared test statistic for this contingency table ssing *scipy*. All you need to do is convert the observed frequencies into a two-dimensional **array**, which is necessary for the scipy function.

```python
import numpy as np
f_obs = np.array([contingency_table.iloc[0][0:6].values,
                  contingency_table.iloc[1][0:6].values])
f_obs
```

```python
from scipy import stats
stats.chi2_contingency(f_obs)[0:3]
```

We see the value of the test statistic, the p-value, and the degrees of freedom here.

Since the p-value is extremely small, we confidently reject the null hypothesis. There is evidence of a relationship between 'sex' and 'hours-per-week'. However, we do not know the direction of this relationship, or which categories are associated with each other. We only now that these two variables are not independent of each other.

## Simple linear regression

Regression analysis searches for relationships among variables. The term *linear* refers to a linear relationship between variables. If we draw this relationship in a two-dimensional space, we get a straight line.

There are several possible Python packages that you can use to run a linear regression. Here, we will use again the package *scipy*. To estimate the slope of the line, we are going to use *linregress* from the *scipy.stats* module.

## from scipy.stats import linregress

We once again use the *tips* data as an example.

```python
import seaborn as sns
tips = sns.load_dataset('tips')
tips.head()
```

There are two types of variables: one dependent variable (the response or the outcome) and one or more independent variables (the predictors). A regression model describes the mean of the outcome variable for given predictor values. In a linear regression model, the (conditional) mean of the outcome variable depends linearly on the predictor values.

In our example, we would like to set up a regression model to predict *tip* using *total_bill* as a predictor.

```python
# Extract the variables
subset = tips.dropna(subset = ['total_bill', 'tip'])  # Linregress cannot handle missing values so drop them
xs = subset['total_bill']
ys = subset['tip']
```

```python
# Run regression with scipy
res = stats.linregress(xs, ys)
res
```

The result is a *linregress* result object, that contains five values: 
    
       - slope: the estimate of the slope of the line that best fits our data; 
       - intercept: the estimate of the intercept of that line;
       - rvalue: the sample correlation coefficient;
       - pvalue: the p-value of the slope (testing the null hypothesis that the slope is zero); 
       - stderr: standard error of the estimate of the slope.
       
The fitted model for the tips at the restaurant as a function of total bill is:

E[tip | total_bill] = 0.92 + 0.10 * total_bill

How do we interpret the slope? For each dollar increment in the total bill, the mean tip increases by 10 cents.

To better understand the slope, we can plot the regression line by using the *seaborn* method *lmplot*: 

```python
sns.lmplot(data = tips,
           x ='total_bill', 
           y ='tip')
```

Another option to do linear regression is the *statsmodels* package. It is a powerful Python package for the estimation of statistical models, performing tests, ... Indeed, in case you want to fit a multiple regression model, you have to use *statsmodels* since *scipy* does not do multiple regression. You can find more information on *statsmodels* on its official website: https://www.statsmodels.org/stable/index.html.

Let us run the same regression model using *statsmodels* to confirm the results.

```python
import statsmodels.api as sm
```

```python
X = tips["total_bill"]
Y = tips["tip"]

X = sm.add_constant(X.values)
X
```

```python
model = sm.OLS(Y, X).fit()
predictions = model.predict(X)

print_model = model.summary()
print(print_model)
```

```python
# dir(model) # gives you all the info present in the fitted model object, you can access them all for exploration
```

```python
print('Adj R squared =', model.rsquared_adj) # for the adjusted r squared
```

```python
print('Predicted values =', model.predict()) # for predicted values
```

Let us test for normality of the residuals using an omnibus test.

```python
import statsmodels.stats.api as sms   #the .stats version of statsmodel gives some extra diagnostics options
test = sms.omni_normtest(model.resid)
test
```
