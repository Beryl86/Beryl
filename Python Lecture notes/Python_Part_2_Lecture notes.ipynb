# Functions

## Functions 


Functions are the most important tool to efficiently code in Python. They are great in case you need to repeat the same or very similar code several times. 

### Defining and Calling a Function

A function is defined by the keyword *def* and it returns the object in the *return* statement. Multiple *return* statements are allowed. Here is an example:

```python
def greet():
    print("Hello! Welcome to Python.")

# Call the function
greet()
```

### Functions with Parameters

You can pass information into a function using **parameters**:

```python
def greet(name):
    print(f"Hello, {name}! Welcome to Python.")

greet("Alice")
greet("Bob")
```

### Functions that Return a Value

A function can also **return** a result using the *return* keyword:

```python
def add(a, b):
    result = a + b
    return result

sum1 = add(3, 5)
sum2 = add(10, 20)

print(sum1)   # 8
print(sum2)   # 30
```

```python
def my_funct(x, y, z = 2):
    if z > 1:
        return z * (x + y)
    else:
        return z / (x + y)     
```

The third argument has been given a default value of 2. If we do not specify a third argument, the default is used.

```python
output = my_funct(2, 3, 0.5)
output
```

```python
my_funct(2, 3)
```

When a function is called, arguments can be provided as **positional arguments** and **keyword arguments**. 

In the following function call, all arguments were provided as positional arguments. The right value is attached to the right variable name based on position: the first one is x, the second one is y, and the third one is z.



```python
my_funct(2, 3, 0.5)
```

In the following function call, all arguments were provided as keyword arguments. Note that the order does not matter. 

```python
my_funct(y = 2, x = 3)
```

Even though I do not recommend it, you can use a combination of positional and keyword arguments. There are some restrictions on what you can do, however, as illustrated below. 

* All positional arguments need to preceed all keyword arguments.
* The positional arguments need to be in the right position.

```python
my_funct(x = 3, y = 2)
```

```python
my_funct(3, y = 2)
```

Here is another example of a function, this time including an *if else* statement:

```python
def nationality(name):
    if name == "John":
        print("I am from Kenya!")
    elif name == "Mary":
        print("I am from the USA!")
    else:
        print(name + ", where are you from?")
nationality("John")
nationality("Mary")
```

```python
nationality("Christina")
```

#  *NumPy*

In general, a Python library is a collection of functions and methods that allows you to perform actions without writing your own code. If you are working with data, some libraries are foundational: *NumPy, pandas, Matplotlib*, ...

*NumPy*, short for Numerical Python, is one of the most important foundational packages for numerical computing in Python. The aim of this chapter is to give an understanding of *NumPy* arrays and array-oriented computation, which will help you use tools with array-oriented semantics, like *pandas*, in a more effective way. 

*NumPy* provides an alternative to the Python list data structure: *NumPy* array. It is similar to a list but it has additional features, such as the ability to perform calculation over entire arrays in an easy and fast way. It is not part of the core Python installation (download from www.python.org), so you have to download it. However, NumPy is part of the "Anaconda distribution" (download from https://anaconda.org).

You can import the *NumPy* package to use it in your Python session; *np* is the standard convention used to import *NumPy*. You can decide to not follow this convention, but we will in this manual. 

```python
import numpy as np
```

## *NumPy* arrays 

We will cover only a small portion of *NumPy* features in this chapter. The first important concept we introduce is an array. To create a *NumPy* array, you can use the *array* function and simply specify a regular Python list as input data.

```python
height_array = np.array([1.55, 1.78, 1.65, 1.67, 1.88, 1.71])  # height of people
height_array
```

```python
weight_array = np.array([52.1, 66.3, 65.8, 55.9, 84.5, 67.3])  # weight of people 
weight_array 
```

If we want to calculate the Body Mass Index (BMI) of this group of people, we can do it as follows:

```python
bmi = weight_array / height_array ** 2
bmi
```

The calculations with arrays were performed element-wise.

If you would have tried to do the same calculation with regular lists, you would get an error!

```python
# bmi calculation with lists 
height = [1.55, 1.78, 1.65, 1.67, 1.88, 1.71]
weight = [52.1, 66.3, 65.8, 55.9, 84.5, 67.3]
weight / height ** 2 # ERROR!
```

Recall that the + operator for lists concatenates the lists. For arrays, however, it computes the element-wise addition.

```python
height + weight # concatenation
```

```python
height_array + weight_array # element-wise addition
```

Besides the fact that NumPy operations tend to act element-wise, there is one other important consideration to keep in mind: *NumPy* assumes that your *NumPy* array only contains values of a single type. If you try to create an array with different types, the resulting *NumPy* array will contain a single type. In the following example, the Boolean and the float are converted to strings:

```python
np.array([7.5, "today", False])
```

In general, you can work with *NumPy* arrays in the same as you can with regular Python lists.

```python
# print the array
height_array
```

```python
# access the first element
height_array[4]
```

There is specific way to subset a *NumPy* array using an array of Booleans. This does not work for lists.

```python
height_array
```

```python
height_array < 1.7
```

```python
height_array[height_array <= 1.67]   #Select only the elements whose condition is True
```

Take a look at the type of *NumPy* array *height* that we created in the previous section.

The output indicates it is a type defined in the *NumPy* package, where *ndarray* stands for "n-dimensional array". 

Our array *height_array* is a one-dimesional array, but is it possible to create two, three, ..., or n-dimensional arrays! You can create 2D arrays from a regular Python list of lists.

```python
array_2d = np.array([[1.55, 1.78, 1.65, 1.67, 1.88, 1.71],
                    [52.1, 66.3, 65.8, 55.9, 84.5, 67.3]])
array_2d
```

We obtain a data structure where each sublist in the list corresponds to a row in the 2-dimensional array. You can obtain the number of rows and columns using the *shape* attribute.

```python
array_2d.shape
```

```python
array_2d[1] # select the second row
```

## Basic Statistics with *Numpy* 

You can generate useful summary statistics about you data with *NumPy*.

```python
Mean=np.mean(height_array) # mean
print(Mean)
Median=np.median(height_array) # median
print(Median)
Variance=np.var(height_array) # variance
print(Variance)
Std=np.std(height_array) # standard deviation 
print(Std)
Min=np.min(height_array) # minimum
print(Min)
Max=np.max(height_array) # maximum
print(Max)
Cor=np.corrcoef(height_array, weight_array) # correlation
print(Cor)
```

We can even compute the mean/median/var/std/min/max of all rows or columns in a 2d array using the axis option. 

```python
array_2d
```

```python
Mean_Wgt_Hgt=np.mean(array_2d) # Computes the overall mean of all values in the entire array (does not make much sense)
print(Mean_Wgt_Hgt)
Mean_Wgt_Hgt_pp=np.mean(array_2d, axis = 0) # mean of weight and height per person- Computes the mean column by column(does not make much sense either)
print(Mean_Wgt_Hgt_pp)
Mean_Hgt_Wgt=np.mean(array_2d, axis = 1) # mean height and mean weight -Computes the mean row by row
print(Mean_Hgt_Wgt)
```
